```python
import os

# Base folder where your processed .pkl files are stored
base_folder = "processed_data"

# List of day strings (can be dynamic)
days = [
    "2020_01_03",
    "2020_01_04",
    "2020_01_05",
    "2020_01_06",
    "2020_01_07",
    "2020_01_08",
]

# Construct dictionary in a safe way
daywise_paths = {day: os.path.join(base_folder, f"AIS_{day}.pkl") for day in days}

# Optional: verify all paths exist
for day, path in daywise_paths.items():
    if not os.path.exists(path):
        print(f"Warning: File does not exist -> {path}")
```
